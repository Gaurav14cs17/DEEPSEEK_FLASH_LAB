## Setup

```python
from pathlib import Path
import sys

PACKAGES = Path.cwd()
while not (PACKAGES / "qwen_common").exists():
    if PACKAGES == PACKAGES.parent:
        raise RuntimeError("Run this notebook from inside deepseek_flash_lab")
    PACKAGES = PACKAGES.parent
sys.path.insert(0, str(PACKAGES))

import torch

torch.manual_seed(0)
```

## Reasoning format

```python
from qwq_mini import QwQMini, preset

prompt = QwQMini.format_reasoning("What is 2 + 2?")
print(prompt)
print(QwQMini.parse_reasoning(" thinking\n2 + 2 = 4\n</think>\nThe answer is 4."))
```

## Forward pass

```python
cfg = preset("demo")
model = QwQMini(cfg)
tokens = torch.randint(0, cfg.vocab_size, (2, 8))
logits, _ = model(tokens)
print("logits:", tuple(logits.shape), "params:", f"{model.num_parameters():,}")
print("thinking:", model.cfg.thinking)
```
