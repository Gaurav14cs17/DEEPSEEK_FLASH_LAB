# Qwen3 Mini

Qwen3 adds QK-Norm and a thinking/non-thinking mode flag.

## Setup

```python
from pathlib import Path
import sys

PACKAGES = Path.cwd()
while not (PACKAGES / "qwen_common").exists():
    if PACKAGES == PACKAGES.parent:
        raise RuntimeError("Run this notebook from inside deepseek_flash_lab")
    PACKAGES = PACKAGES.parent
sys.path.insert(0, str(PACKAGES))

import torch

torch.manual_seed(0)
```

## QK-Norm

```python
from qwen3_mini import Qwen3Model, preset

cfg = preset("demo")
print("use_qk_norm:", cfg.use_qk_norm)
model = Qwen3Model(cfg)
tokens = torch.randint(0, cfg.vocab_size, (2, 8))
logits, _ = model(tokens)
print("logits:", tuple(logits.shape), "params:", f"{model.num_parameters():,}")
```

## Thinking mode

This is a prompt/decoding policy, not a different weight layout.

```python
model.set_thinking(True)
print(model.thinking_prompt("What is 2 + 2?"))

model.set_thinking(False)
print(model.thinking_prompt("What is 2 + 2?"))
```
