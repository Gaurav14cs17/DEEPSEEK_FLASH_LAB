# Profiling

Time a forward pass.


```python
import sys
from pathlib import Path

PACKAGES = Path.cwd()
while not (PACKAGES / "deepseek_flash_v4_mini").exists():
    if PACKAGES == PACKAGES.parent:
        raise RuntimeError("Run this notebook from inside deepseek_flash_lab")
    PACKAGES = PACKAGES.parent
sys.path.insert(0, str(PACKAGES))
import torch
from deepseek_flash_v4_mini.model import DeepSeekFlashV4Mini, preset
from deepseek_flash_v4_mini.optimization.profiling import Timer
m=DeepSeekFlashV4Mini(preset('demo')); x=torch.randint(0,m.cfg.vocab_size,(2,32))
t=Timer()
with t.track('forward'):
    m(x,trace=False)
print(t.report())
```
